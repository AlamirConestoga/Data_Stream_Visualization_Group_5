```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import os
```

```python
#load linked data
df = pd.read_csv("data/RMBR4-2_export_test_linked.csv")
```

```python
# Ensure the 'Time' column is parsed as datetime and sort ascending
df["Time"] = pd.to_datetime(df["Time"])
df_sorted = df.sort_values(by="Time", ascending=True).reset_index(drop=True)

df_sorted.head()
```

```python
# Filter for active rows (where Axis #1 > 0) and sort ascending
df_active_sorted = (
    df[df["Axis #1"] > 0]
    .sort_values(by="Axis #1", ascending=True)
    .reset_index(drop=True)
)

# Calculate Median & Mean on non-zero active observations
print("Active Mean:", df_active_sorted["Axis #1"].mean())
print("Active Median:", df_active_sorted["Axis #1"].median())
```
